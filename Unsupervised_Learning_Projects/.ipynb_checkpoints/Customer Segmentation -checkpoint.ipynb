```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
customer_data = pd.read_csv("Mall_Customers.csv")
```

```python
customer_data.head(5)
```

```python
customer_data.isnull().sum()
```

```python
X = customer_data.iloc[:,[3,4]].values
```

```python
print(X)
```

```python
wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)   # <-- Correct
```

```python
sns.set()
plt.plot(range(1,11), wcss)
plt.title("Elbow Plot")
plt.xlabel("Number of Cluster")
plt.show()
```

```python
 kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
```

```python
print(X)
```

```python
y= kmeans.fit_predict(X)
```

```python
print(y)
```

```python
plt.scatter(X[y == 0, 0], X[y == 0, 1], label='Cluster 1')
plt.scatter(X[y == 1, 0], X[y == 1, 1], label='Cluster 2')
plt.scatter(X[y == 2, 0], X[y == 2, 1], label='Cluster 3')
plt.scatter(X[y == 3, 0], X[y == 3, 1], label='Cluster 4')
plt.scatter(X[y == 4, 0], X[y == 4, 1], label='Cluster 5')

plt.scatter(kmeans.cluster_centers_[:, 0],
            kmeans.cluster_centers_[:, 1],
            s=200, c='black', marker='X', label='Centroids')

plt.legend()
plt.show()
```
